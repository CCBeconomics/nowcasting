```python
import io, os, re, urllib3, requests, datetime, shutil, glob 
import pandas as pd
import numpy as np


d = os.getcwd()+"/"
PATH_RAW = f"{d}/../source"
PATH_FORMAT = f"{d}/../data"

# ----------------------------------
# Append Google Trends

csv_files = glob.glob(os.path.join(PATH_RAW, "*multi*.csv"))
if len(csv_files)>0:
    df0 = pd.DataFrame()
    for csv in csv_files :
        dfi = pd.read_csv(csv, skiprows=1)
        dfi.rename(columns = {"Month" : 'date'}, inplace=True)
        dfi.date = pd.to_datetime(dfi['date'])
        dfi.columns = dfi.columns.astype(str).str.lower().str.strip()
        dfi.columns = [re.sub(r'\W+', '_', col).strip('_') for col in dfi.columns]
        dfi.set_index('date', inplace=True)
    
        df0 = pd.concat( [ df0, dfi], axis=1)
    df0.columns = ["gt_" + col for col in df0.columns]
    
    df0.to_csv(f'{PATH_RAW}/gtrends.csv')
    print(f'Last updated: {df0.index.max()}')
    
    for file in csv_files :
        os.remove(file)
```

```python
# Rename columns with the specified substitutions
df0.columns = df0.columns.str.replace('gtrends_', 'gt_', regex=False) \
                       .str.replace('_bahamas', '_bs', regex=False) \
                       .str.replace('_united_states', '_us', regex=False) \
                       .str.replace('_worldwide_', '_', regex=False) \
                       .str.replace('_bs_', '_bahamas_', regex=False)
```

```python
df0.to_csv(f'{PATH_RAW}/gtrends.csv')
print(f'Last updated: {df0.index.max()}')
```
