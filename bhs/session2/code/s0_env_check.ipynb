## Nowcast Environment

Follow these steps:

- Create the ```nwcst``` environment

```conda create -n nwcst python=3.11 pytorch pip numpy pandas matplotlib seaborn scikit-learn tqdm python-dateutil geopandas  jupyterlab -y```

If there are errors with pytorch. Do this:

```conda create -n nwcst python=3.11 pip numpy pandas matplotlib seaborn scikit-learn tqdm python-dateutil geopandas  jupyterlab -y```

Then:

```conda activate nwcst```

If there are errors with pytorch, we solve them now:

```conda clean --packages --tarballs```

```conda install pytorch torchvision torchaudio cpuonly -c pytorch```

- Add additional packages

```conda install pip pmdarima -y```

```pip install nowcast_lstm```

```conda install -c conda-forge pypdf2 camelot-py -y```

(May be needed: ```conda install -y -c conda-forge cryptography openssl cffi pdfminer.six``` or ```conda install -y -c conda-forge pdfminer.six=20220319```)
